# Normal Neural Network
#        │
#        ▼
# No memory
#        │
#        ▼
# RNN
#        │
#        ▼
# Has memory
#        │
#        ├── Slow (sequential)
#        ├── Forgets long sequences
#        ▼
# LSTM
#        │
#        ▼
# Better memory
#        │
#        ├── Still sequential
#        ├── Still hard for very long contexts
#        ▼
# Transformer
#        │
#        ▼
# No long memory chain
# Every word attends to every other word
# Parallel processing
# Better long-range understanding

```python
from transformers import pipeline

classifier = pipeline("sentiment-analysis")

print(classifier("I machine learn"))
```

```python
# print(classifier.model)
print(classifier.model.config)
```

```python
for name, module in classifier.model.named_modules():
    print(name)
```
