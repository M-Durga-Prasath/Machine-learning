```python
import time
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.metrics.pairwise import euclidean_distances
```

```python
models = {
    "all-MiniLM-L6-v2",
    "all-mpnet-base-v2"
}

loaded_models = {}

for name in models:
    start = time.time()

    model = SentenceTransformer(name)

    end = time.time()

    loaded_models[name] = model
    print("=" * 50)
    print(name)
    print(f"Load time: {end-start:.2f} sec")
    print("Embedding dimension:", model.get_embedding_dimension())
```

```python
corpus = [

    "Machine learning allows computers to learn from data.",

    "Deep learning uses neural networks with many layers.",

    "Docker packages applications inside containers.",

    "Containers are lightweight compared to virtual machines.",

    "YOLO is an object detection model.",

    "Object detection predicts bounding boxes.",

    "I enjoy building AI applications.",

    "Python is my primary programming language.",

    "RAG combines retrieval with language models.",

    "Embeddings represent semantic meaning.",

    "I enjoy reading books on machine learning.",

    "Vector databases store embeddings.",

    "Cosine similarity compares vector direction.",

    "I like eating pizza on weekends."

]

```

```python
for name, model in loaded_models.items():

    print(f"\n{name}")

    embeddings = model.encode(corpus)

    print("Embedding Shape:", embeddings.shape)
```

```python
pairs = [

(
"Deep learning is a subset of machine learning.",
"Neural networks learn patterns."
),

(
"Deep learning is a subset of machine learning.",
"I like eating pizza."
),

(
"Docker uses containers.",
"Containers package applications."
),

(
"YOLO detects objects.",
"I love reading novels."
)

]
```

```python
for model_name, model in loaded_models.items():

    print("\n", "="*60)
    print(model_name)

    for s1, s2 in pairs:

        emb = model.encode([s1,s2])

        cosine = cosine_similarity(
            [emb[0]],
            [emb[1]]
        )[0][0]

        euclidean = euclidean_distances(
            [emb[0]],
            [emb[1]]
        )[0][0]

        print()

        print(s1)
        print(s2)

        print(f"Cosine Similarity : {cosine:.4f}")
        print(f"Euclidean Distance: {euclidean:.4f}")
```

```python
documents = [

"Deep learning uses neural networks.",

"Docker containers simplify deployment.",

"YOLO performs object detection.",

"Embeddings represent sentence meaning.",

"I enjoy pizza every Friday."

]
```

```python
query = "How do neural networks learn?"

doc_embeddings = model.encode(documents)

query_embedding = model.encode(query)


scores = cosine_similarity(
    [query_embedding],
    doc_embeddings
)[0]

scores
```

```python
ranking = sorted(
    zip(documents,scores),
    key=lambda x:x[1],
    reverse=True
)

for doc,score in ranking:

    print(f"{score:.3f} -> {doc}")
```

```python
import asyncio
import time


async def generate_embedding(sentence):
    print(f"Generating embedding for: {sentence}")
    await asyncio.sleep(1)
    return f"Embedding({sentence})"


async def sequential():
    sentences = [
        "Machine Learning",
        "Docker",
        "YOLO"
    ]

    results = []

    for sentence in sentences:
        result = await generate_embedding(sentence)
        results.append(result)

    return results


start = time.perf_counter()

res = await sequential()

end = time.perf_counter()

print(f"Time: {end-start:.2f} seconds")
res 
```
